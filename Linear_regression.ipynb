# Linear Regression Implementation
This notebook demonstrates a simple linear regression model using scikit-learn

```python
# Import necessary libraries
import numpy as np
import matplotlib.pyplot as plt
import joblib  # Modern way to save/load sklearn models
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_squared_error
```

```python
# Create dataset
# Independent variable (Hours studied)
X = np.array([2, 4, 6, 8, 10]).reshape(-1, 1)

# Dependent variable (Scores)
y = np.array([50, 60, 70, 80, 90])
```

```python
# Visualize the data
plt.scatter(X, y, color='blue', label='Actual Data')
plt.xlabel('Hours Studied')
plt.ylabel('Scores')
plt.title('Hours Studied vs Scores')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# Create and train the Linear Regression model
model = LinearRegression()
model.fit(X, y)

print("Model training completed!")
```

```python
# Display model parameters
print(f"Coefficient (Slope): {model.coef_[0]:.2f}")
print(f"Intercept: {model.intercept_:.2f}")
print(f"\nLinear Equation: y = {model.coef_[0]:.2f}x + {model.intercept_:.2f}")
```

```python
# Make predictions
y_pred = model.predict(X)

# Calculate model performance metrics
r2 = r2_score(y, y_pred)
mse = mean_squared_error(y, y_pred)
rmse = np.sqrt(mse)

print(f"R² Score: {r2:.4f}")
print(f"Mean Squared Error (MSE): {mse:.4f}")
print(f"Root Mean Squared Error (RMSE): {rmse:.4f}")
```

```python
# Visualize the regression line
plt.scatter(X, y, color='blue', label='Actual Data')
plt.plot(X, y_pred, color='red', linewidth=2, label='Regression Line')
plt.xlabel('Hours Studied')
plt.ylabel('Scores')
plt.title('Linear Regression: Hours Studied vs Scores')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# Make a prediction for a new value
hours_studied = [[7]]  # Predicting score for 7 hours of study
predicted_score = model.predict(hours_studied)

print(f"Predicted score for {hours_studied[0][0]} hours of study: {predicted_score[0]:.2f}")
```

```python
# Visualize prediction with the new data point
plt.scatter(X, y, color='blue', label='Training Data')
plt.plot(X, y_pred, color='red', linewidth=2, label='Regression Line')
plt.scatter(hours_studied, predicted_score, color='green', s=100, marker='*', 
            label=f'Prediction ({hours_studied[0][0]}h, {predicted_score[0]:.1f})', zorder=5)
plt.xlabel('Hours Studied')
plt.ylabel('Scores')
plt.title('Linear Regression with Prediction')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# Save the trained model using joblib
# joblib is more efficient for sklearn models than pickle
joblib.dump(model, 'model_joblib.pkl')
```

```python
# Verify the model was saved
print("Model saved successfully as 'model_joblib.pkl'")
```

```python
# Load the saved model using joblib
loaded_model = joblib.load('model_joblib.pkl')
print("Model loaded successfully!")



```

```python
# Make predictions using the loaded model
test_input = [[7]]  # Testing with 7 hours
prediction = loaded_model.predict(test_input)
print(f"Prediction for {test_input[0][0]} hours: {prediction[0]:.2f}")
```
