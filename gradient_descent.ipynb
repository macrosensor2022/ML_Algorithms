# Gradient Descent Implementation
This notebook demonstrates gradient descent algorithm for linear regression

**IMPORTANT**: Please run all cells in order from top to bottom. If you encounter errors, restart the kernel (Kernel → Restart Kernel) and run all cells again.

```python
# Import required libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib import animation
from sklearn.linear_model import LinearRegression
```

## Example 1: Simple Dataset

```python
def gradient_descent(x, y, learning_rate=0.08, iterations=10000):
    # Initialize parameters
    m_current = b_current = 0
    n = len(x)
    
    # Store history for visualization
    m_values = [m_current]
    b_values = [b_current]
    
    for i in range(iterations):
        # Calculate predictions
        y_predicted = m_current * x + b_current
        
        # Calculate cost (Mean Squared Error)
        cost = (1/n) * sum([val**2 for val in (y - y_predicted)])
        
        # Calculate gradients
        md = -(2/n) * sum(x * (y - y_predicted))
        bd = -(2/n) * sum(y - y_predicted)
        
        # Update parameters
        m_current = m_current - learning_rate * md
        b_current = b_current - learning_rate * bd
        
        # Store values
        m_values.append(m_current)
        b_values.append(b_current)
        
        # Print progress every 1000 iterations
        if i % 1000 == 0:
            print(f"m {m_current:.4f}, b {b_current:.4f}, cost {cost:.6f}, iteration {i}")
    
    return m_values, b_values
```

```python
# Simple training data (y = 2x + 3)
x = np.array([1, 2, 3, 4, 5])
y = np.array([5, 7, 9, 11, 13])

# Run gradient descent
m_values, b_values = gradient_descent(x, y)

print("\n=" * 50)
print("FINAL RESULTS")
print("=" * 50)
print(f"Slope (m): {m_values[-1]:.4f}")
print(f"Intercept (b): {b_values[-1]:.4f}")
print(f"Expected: m=2, b=3 (since y = 2x + 3)")
```

```python
# Visualize the final fitted line
plt.figure(figsize=(10, 6))
plt.scatter(x, y, color='blue', label='Data Points', s=100, zorder=5)

# Plot the final fitted line
final_m = m_values[-1]
final_b = b_values[-1]
y_fitted = final_m * x + final_b
plt.plot(x, y_fitted, 'r-', linewidth=3, label=f'Fitted Line: y = {final_m:.4f}x + {final_b:.4f}')

plt.xlim(0, 6)
plt.ylim(0, 16)
plt.xlabel('X', fontsize=12)
plt.ylabel('Y', fontsize=12)
plt.title('Gradient Descent - Final Result', fontsize=14, fontweight='bold')
plt.legend(fontsize=11)
plt.grid(True, alpha=0.3)
plt.show()
```

## Example 2: Test Scores Dataset
Finding correlation between math and CS scores using gradient descent

```python
# Load the dataset
df = pd.read_csv('test_scores.csv')
print("Dataset:")
print(df.head())

# Extract data
x_test = np.array(df.math)
y_test = np.array(df.cs)

# Compare with sklearn
sklearn_model = LinearRegression()
sklearn_model.fit(x_test.reshape(-1, 1), y_test)
print(f"\nSklearn - Coefficient: {sklearn_model.coef_[0]:.4f}, Intercept: {sklearn_model.intercept_:.4f}")
```

```python
# IMPORTANT: Lower learning rate for larger values
# Math scores range from 56-92, so we need a much smaller learning rate
# Using learning_rate=0.0001 and more iterations for convergence
m_values_test, b_values_test = gradient_descent(x_test, y_test, learning_rate=0.0001, iterations=100000)

# Get final values
m_final = m_values_test[-1]
b_final = b_values_test[-1]

print("\n=" * 50)
print("GRADIENT DESCENT RESULTS")
print("=" * 50)
print(f"Coefficient: {m_final:.4f}")
print(f"Intercept: {b_final:.4f}")
print(f"\nComparison with sklearn:")
print(f"Difference in coefficient: {abs(m_final - sklearn_model.coef_[0]):.4f}")
print(f"Difference in intercept: {abs(b_final - sklearn_model.intercept_):.4f}")
```

```python
# Visualize the fitted line for test scores
plt.figure(figsize=(10, 6))
plt.scatter(x_test, y_test, color='blue', label='Data Points', s=100, zorder=5)

# Plot gradient descent line
y_fitted_test = m_final * x_test + b_final
plt.plot(x_test, y_fitted_test, 'r-', linewidth=3, 
         label=f'Gradient Descent: y = {m_final:.4f}x + {b_final:.4f}')

# Plot sklearn line for comparison
y_sklearn = sklearn_model.predict(x_test.reshape(-1, 1))
plt.plot(x_test, y_sklearn, 'g--', linewidth=2, alpha=0.7,
         label=f'Sklearn: y = {sklearn_model.coef_[0]:.4f}x + {sklearn_model.intercept_:.4f}')

plt.xlabel('Math Score', fontsize=12)
plt.ylabel('CS Score', fontsize=12)
plt.title('Math vs CS Scores - Gradient Descent', fontsize=14, fontweight='bold')
plt.legend(fontsize=10)
plt.grid(True, alpha=0.3)
plt.show()
```
