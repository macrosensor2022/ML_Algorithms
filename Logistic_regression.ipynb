# Logistic Regression Implementation
This notebook demonstrates binary classification using Logistic Regression

```python
# Import necessary libraries
import numpy as np
import pandas as pd
from sklearn.linear_model import LogisticRegression
import matplotlib.pyplot as plt
```

```python
# Create dataset
# Independent variable (Hours studied) - reshaped to 2D array for sklearn
x = np.array([1, 2, 3, 4, 5, 6]).reshape(-1, 1)

# Dependent variable (Binary outcome: 0 = Fail, 1 = Pass)
y = np.array([0, 0, 0, 1, 1, 1])
```

```python
# Visualize the training data
plt.scatter(x, y, color='blue', label='Training Data')
plt.xlabel('Hours Studied')
plt.ylabel('Pass/Fail (0 or 1)')
plt.title('Student Study Hours vs Pass/Fail Outcome')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# Create and train the Logistic Regression model
model = LogisticRegression()
model.fit(x, y)

print("Model training completed successfully!")
```

```python
# Display model parameters
print("=" * 50)
print("MODEL PARAMETERS")
print("=" * 50)
print(f"Coefficient (Weight): {model.coef_[0][0]:.4f}")
print(f"Intercept (Bias): {model.intercept_[0]:.4f}")

# Make a prediction for 4 hours of study
hours = [[4]]
probability = model.predict_proba(hours)

print("\n" + "=" * 50)
print("PREDICTION FOR 4 HOURS OF STUDY")
print("=" * 50)
print(f"Probability of Passing: {probability[0][1]:.4f} ({probability[0][1]*100:.2f}%)")
print(f"Probability of Failing: {probability[0][0]:.4f} ({probability[0][0]*100:.2f}%)")
```

```python
# Make final binary prediction (0 or 1)
prediction = model.predict(hours)

print("=" * 50)
print(f"Final Prediction: {'PASS' if prediction[0] == 1 else 'FAIL'}")
print("=" * 50)
```

```python
# Visualize the Sigmoid Curve and Decision Boundary
prob = model.predict_proba(hours)[0][1]

# Generate test points for smooth sigmoid curve
x_test = np.linspace(0, 7, 300).reshape(-1, 1)
y_prob = model.predict_proba(x_test)[:, 1]

# Create comprehensive visualization
plt.figure(figsize=(10, 6))
plt.scatter(x, y, label="Training Data (0=Fail, 1=Pass)", color='blue', s=100, zorder=3)
plt.plot(x_test, y_prob, label="Sigmoid Curve (Probability)", color='green', linewidth=2)
plt.axhline(0.5, linestyle="--", color='orange', label="Decision Boundary (0.5)", linewidth=2)
plt.scatter(hours, prob, color="red", s=150, marker='*', 
            label=f"Prediction for {hours[0][0]}h (P={prob:.2f})", zorder=5)
plt.vlines(hours, 0, prob, linestyle="dotted", color='red', linewidth=2)
plt.xlabel("Hours Studied", fontsize=12)
plt.ylabel("Probability of Passing", fontsize=12)
plt.title("Logistic Regression: Decision Boundary Visualization", fontsize=14, fontweight='bold')
plt.legend(loc='best')
plt.grid(True, alpha=0.3)
plt.ylim(-0.1, 1.1)
plt.show()
```
