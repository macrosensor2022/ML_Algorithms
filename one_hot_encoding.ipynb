```python
import pandas as pd
df = pd.read_csv(r"homeprices.csv")
df.head(20)
```

```python
dummies = pd.get_dummies(df.town)
dummies.head(20)



```

```python
merged = pd.concat([df,dummies],axis = 'columns')
merged.head()
```

```python
#Remove the town column to avoid producing dummy variable trap that messes up the model

final = merged.drop(['town','west windsor'],axis='columns')
final.head()
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
```

```python
model
```

```python
x= final.drop('price',axis='columns')
x
```

```python
y= final.price
```

```python
model.fit(x,y)
```

```python
model.predict([[3400,0,1]])
```

```python
# Key Takeaways - One-Hot Encoding

## Summary of Methods Covered:

| Method | Pros | Cons |
|--------|------|------|
| **pd.get_dummies()** | Simple, quick for exploration | Not ideal for production pipelines |
| **LabelEncoder** | Simple integer encoding | Implies ordinal relationship (problematic for linear models) |
| **ColumnTransformer + OneHotEncoder** | Production-ready, integrates with sklearn pipelines | Slightly more verbose |

## Important Concepts:

1. **Dummy Variable Trap**: Always drop one category to avoid perfect multicollinearity
2. **Reference Category**: The dropped category becomes the baseline for interpretation
3. **Feature vs Target**: Always separate X (features) from y (target) before training

## When to Use What:
- **Linear Models**: Use One-Hot Encoding (avoid Label Encoding)
- **Tree-based Models**: Can use either (trees handle ordinal encoding well)
- **Production**: Use ColumnTransformer for reproducible pipelines

*Last updated: February 21, 2026*
```
